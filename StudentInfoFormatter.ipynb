```python
# This is a program that asks for student details and displays formatted output.
# Standards: The program is expected to use at least two string manipulation functions or methods and should run without syntax errors.

first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
favorite_subject = input("Enter favorite subject: ")

# Process the inputs and format them.
full_name = (first_name.strip() + " " + last_name.strip()).title()
name_code = full_name.upper()
section_formatted = section.strip().upper()
favorite_subject_formatted = favorite_subject.strip().title()
character_count = len(full_name)

print("\nFormatted Output:")
print("Name:", full_name)
print("Name Code:", name_code)
print("Section:", section_formatted)
print("Favorite Subject:", favorite_subject_formatted)
print("Character Count:", character_count)
```
